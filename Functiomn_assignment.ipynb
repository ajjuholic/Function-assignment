4. What is the purpose of the `return` statement in a Python function?
Ans -- The return statement in a Python function serves two primary purposes:

1.  Terminating the Function: When the return statement is encountered within a function, it immediately terminates the execution of the function, regardless of whether there's any remaining code within the function body.

2.  Returning a Value: The return statement can be used to send a value back to the caller of the function. This value can be a single variable, a tuple of values, or any other valid Python object.



```python
# Example return statement
def add_numbers(x, y):
    """This function adds two numbers and returns the result."""
    sum = x + y
    return sum              # Returning the sum of the two numbers
result = add_numbers(5, 3)  # Calling the function and assigning the returned value
print(result)                # To see the output, run the code.
```

6. Explain the concept of generators in Python and how they are defined.
Ans --
      Generators :

                  Generators are a special type of function in Python that produce a sequence of values one at a time instead of computing them all at once and storing them in memory. They are defined using the `yield` keyword instead of `return`.




      How Generators Work :

         *  When a generator function is called, it doesn't execute the function body immediately. Instead, it returns a generator object.

         * When you iterate over the generator object (e.g., using a for loop or `next()`), the function body is executed until it encounters a `yield` statement.

         * The `yield` statement suspends the function's execution and returns the yielded value.   

         * The next time you request a value from the generator, it resumes execution from where it left off and continues until it encounters another `yield` or the end of the function.   


      Defining Generators

      Generators are defined using the `def` keyword like regular functions, but they use the `yield` keyword to return values.         

```python
# Genrators
def my_generator(n):
    for i in range(n):
        yield i * i

# Create a generator object
gen = my_generator(5)

# Iterate over the generator
for value in gen:
    print(value)  # To see the output, run the code.

```

7. What are the advantages of using generators over regular functions?
Ans -- Advantages of Generators

1.  Memory Efficiency: Generators are memory-efficient because they only    
  compute and store one value at a time. In contrast, regular functions compute and store all values in memory before returning them. This can be a significant advantage when dealing with large datasets or infinite sequences.

2.  Lazy Evaluation: Generators produce values on demand, which means they only
   generate the next value when it's requested. This is known as lazy evaluation. Regular functions, on the other hand, compute all values upfront. Lazy evaluation can be beneficial when you don't need all the values at once or when generating values is computationally expensive.

3.  Improved Readability: Generators can make code more concise and readable,
   especially for tasks involving sequences of values. The `yield `keyword clearly indicates that the function is a generator, and the logic for generating values is often simpler than in regular functions.

```python
# Example : Regular finction

def fibonacci_list(n):
    """Returns a list of Fibonacci numbers up to n."""
    fib_list = [0, 1]
    while fib_list[-1] < n:
        fib_list.append(fib_list[-1] + fib_list[-2])
    return fib_list

# Get the first 10 Fibonacci numbers
fib_numbers = fibonacci_list(10)
print(fib_numbers)  # To see the output, run the code.
```

```python
# Example : Genrator Function :
def fibonacci_generator(n):
    """Generates Fibonacci numbers up to n."""
    a, b = 0, 1
    while a < n:
        yield a
        a, b = b, a + b

# Get the first 10 Fibonacci numbers
for fib_number in fibonacci_generator(10):
    print(fib_number)  # To see the output, run the code.
```

8. What is a lambda function in Python and when is it typically used?
Ans -- Lambda Functions

A lambda function is a small, anonymous function defined using the `lambda `keyword. It can have any number of arguments but can only have one expression. Lambda functions are often used for short, simple operations that don't require a full function definition using `def`.

Syntax:
lambda arguement : expression

    * arguments: The arguments passed to the lambda function, similar to function parameters.
    * expression: The single expression that the lambda function evaluates and returns.

 When to Use Lambda Functions
 Lambda function are typically used in the following scenrios :

  * Short, simple operations: When you need a function for a quick, one-time operation, a lambda function can be more concise than a regular function.

  * As arguments to higher-order functions: Higher-order functions are functions that take other functions as arguments. Lambda functions are often used as arguments to functions like `map`, `filter`, and `reduce`.
  * Creating closures: Lambda functions can be used to create closures, which are functions that remember the values of variables from their enclosing scope.   

```python
# Example 1: Simple lambda function to add two numbers
add = lambda x, y: x + y
result = add(5, 3)
print(result)  # To see the output, run the code.

# Example 2: Using lambda with map to square a list of numbers
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x * x, numbers))
print(squared_numbers)  # To see the output, run the code.

# Example 3: Using lambda with filter to get even numbers from a list
numbers = [1, 2, 3, 4, 5]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)  # To see the output, run the code.
```

9. Explain the purpose and usage of the `map()` function in Python.
ans --  The `map()` function applies a given function to each item of an iterable (e.g., a list, tuple, or string) and returns an iterator that yields the results. It essentially "maps" the function to each element of the iterable.

syntax
map(function, iterable, .. .)

`function`: The function to apply to each item of the iterable.
`iterable`: The iterable containing the items to be processed.
`...`: Optional, additional iterables. If provided, the function must accept the same number of arguments as the number of iterables.

Usage
The `map()` function is commonly used in the following scenarios :
    
      * Applying a function to each element of a list :



  Benefits of using `map()`

      *  Conciseness: It provides a concise way to apply a function to each element of an iterable.
      *  Efficiency: It can be more efficient than using a loop for large iterables.
      *  Functional Programming: It promotes a functional programming style by avoiding explicit loops.

   Important Considerations

Returns an iterator: The `map()` function returns an iterator, so you need to convert it to a list or tuple if you want to access the results directly.
Lazy Evaluation: The function is applied to each item of the iterable only when the iterator is consumed (e.g., using a loop or `list()`).
  


```python
# Applying a function to each element of a list :
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x * x, numbers))
print(squared_numbers)  # Output: [1, 4, 9, 16, 25]
```

```python
# converting data types :

strings = ["1", "2", "3"]
numbers = list(map(int, strings))
print(numbers)  # Output: [1, 2, 3]
```

```python
# processing multiple iterable in parallel :
list1 = [1, 2, 3]
list2 = [4, 5, 6]
result = list(map(lambda x, y: x + y, list1, list2))
print(result)
```

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

Ans -- `map()`

      Purpose: Applies a function to each item of an iterable and returns an iterator with the results.
      Syntax: `map(function, iterable, ...)`

      `reduce()`

      Purpose: Applies a function cumulatively to the items of an iterable, reducing it to a single value.
      Syntax: reduce(function, iterable[, initializer])
      Requires Import: from `functools import reduce`


      `filter()`

      Purpose: Filters elements from an iterable based on a function's truthiness. It returns an iterator with elements that satisfy the condition.
      Syntax:` filter(function, iterable)`

```python
# Example : MAP
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x * x, numbers))
print(squared_numbers)  # Output: [1, 4, 9, 16, 25]
```

```python
# Example : reduce ()
from functools import reduce
numbers = [1, 2, 3, 4, 5]
product = reduce(lambda x, y: x * y, numbers)
print(product)  # Output: 120
```

```python
# Example : filter ()
numbers = [1, 2, 3, 4, 5]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)  # Output: [2, 4]
```
